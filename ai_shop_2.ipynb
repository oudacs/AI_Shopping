```python
# ==========================================================
# 🚀 CELL 2: STEP 3 & 4 (YOLO11 DETECTION & SMART CROP)
# ==========================================================

print("⏳ جاري تحميل محرك الرؤية (YOLO11)...")
try:
    yolo_model = YOLO(str(BEST_MODEL_PATH))
    print("✅ تم تحميل الموديل بنجاح.")
except Exception as e:
    print(f"⚠️ خطأ في تحميل الموديل، يرجى التأكد من وجود best.pt: {e}")

def detect_and_crop(image_path, conf_threshold=0.2): # Lowered threshold for better recall
    print("👁️ جاري فحص الصورة وقص المنتج...")
    img = cv2.imread(str(image_path))
    if img is None: return None, "Error reading image"

    results = yolo_model.predict(source=img, conf=conf_threshold, verbose=False)
    boxes = results[0].boxes if results else None

    if boxes is not None and len(boxes) > 0:
        best_box = max(boxes, key=lambda b: float(b.conf))
        x1, y1, x2, y2 = map(int, best_box.xyxy[0].tolist())

        # More aggressive padding to capture surrounding text labels
        pad = 25
        h, w = img.shape[:2]
        x1, y1 = max(0, x1 - pad), max(0, y1 - pad)
        x2, y2 = min(w, x2 + pad), min(h, y2 + pad)

        cropped_img = img[y1:y2, x1:x2]
        product_class = yolo_model.names[int(best_box.cls)]
        return cropped_img, product_class

    print("⚠️ لم يتم اكتشاف منتج واضح، سيتم تمرير الصورة كاملة.")
    return img, "Unknown"
```

```python
import os
import cv2
import numpy as np
import easyocr
from pathlib import Path

print("⏳ جاري إعداد محرك EasyOCR المطور بفلترة صور متقدمة...")

ROOT_DIR = Path("/content/drive/MyDrive/NTI tasks/AI Shopping Assistant")
DRIVE_OCR_MODELS = ROOT_DIR / "models/easyocr_weights"
DRIVE_OCR_MODELS.mkdir(parents=True, exist_ok=True)

ocr_reader = easyocr.Reader(
    ['ar', 'en'],
    gpu=True,
    model_storage_directory=str(DRIVE_OCR_MODELS),
    download_enabled=True,
    verbose=False
)

class UltimateOCR:
    def __init__(self, reader):
        self.reader = reader

    def _preprocess(self, img):
        # 1. تحويل لرمادي
        gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
        # 2. زيادة التباين (Sharpening)
        kernel = np.array([[-1,-1,-1], [-1,9,-1], [-1,-1,-1]])
        sharpened = cv2.filter2D(gray, -1, kernel)
        # 3. إزالة الضوضاء
        denoised = cv2.fastNlMeansDenoising(sharpened, None, 10, 7, 21)
        return denoised

    def _read_image(self, img, text_thresh=0.2):
        detections = self.reader.readtext(img, text_threshold=text_thresh, low_text=0.2)
        valid_texts = []
        for bbox, text, conf in detections:
            clean = str(text).strip()
            if len(clean) > 1:
                y_center = (bbox[0][1] + bbox[2][1]) / 2
                valid_texts.append({'text': clean, 'y': y_center})

        if not valid_texts:
            return None

        valid_texts = sorted(valid_texts, key=lambda x: x['y'])
        return " \n ".join([item['text'] for item in valid_texts])

    def extract_text(self, img):
        print("📝 جاري الفحص المطور للعلبة بـ 5 تكتيكات ذكية...")

        # 🔵 التكتيك الأول: الصورة الأصلية
        result = self._read_image(img, text_thresh=0.15)
        if result: return result

        # 🔵 التكتيك الثاني: المعالجة المتقدمة (Sharpening & Denoising)
        print("🔄 التكتيك الثاني: تحسين التباين وإزالة الضوضاء...")
        processed = self._preprocess(img)
        result = self._read_image(processed)
        if result: return result

        # 🔵 التكتيك الثالث: تكبير الصورة (Upscaling)
        print("🔄 التكتيك الثالث: تكبير النصوص الصغيرة...")
        upscaled = cv2.resize(img, None, fx=2, fy=2, interpolation=cv2.INTER_CUBIC)
        result = self._read_image(upscaled)
        if result: return result

        # 🔵 التكتيك الرابع: تباين عالي (CLAHE)
        print("🔄 التكتيك الرابع: تطبيق فلتر التباين الموضعي...")
        gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
        clahe = cv2.createCLAHE(clipLimit=3.0, tileGridSize=(8,8))
        contrast = clahe.apply(gray)
        result = self._read_image(contrast)
        if result: return result

        # 🔵 التكتيك الخامس: الدوران
        print("🔄 التكتيك الخامس: محاولة القراءة بزوايا مختلفة...")
        for angle in [cv2.ROTATE_90_CLOCKWISE, cv2.ROTATE_90_COUNTERCLOCKWISE]:
            rotated = cv2.rotate(img, angle)
            result = self._read_image(rotated)
            if result: return result

        return "لم يتم العثور على نصوص واضحة."

ocr_engine = UltimateOCR(ocr_reader)
print("🎉 تم تحديث محرك OCR بنجاح مع دعم المعالجة الصورية المتقدمة!")
```

```python
from sentence_transformers import SentenceTransformer
import faiss
import json
import os
import numpy as np
from pathlib import Path

print("⏳ جاري إعداد محرك البحث المعرفي (RAG) وتأمين الأوزان...")

ROOT_DIR = Path("/content/drive/MyDrive/NTI tasks/AI Shopping Assistant")
DB_DIR = ROOT_DIR / "vector_db"
EMBEDDING_MODEL_DIR = ROOT_DIR / "models/embedding_model"
FAISS_INDEX_PATH = DB_DIR / "products_faiss.index"
METADATA_PATH = DB_DIR / "products_metadata.json"

if not EMBEDDING_MODEL_DIR.exists():
    embedder = SentenceTransformer('paraphrase-multilingual-MiniLM-L12-v2')
    embedder.save(str(EMBEDDING_MODEL_DIR))
else:
    embedder = SentenceTransformer(str(EMBEDDING_MODEL_DIR))

class SmartSearchEngine:
    def __init__(self, embedder, index_path, metadata_path):
        self.embedder = embedder
        self.index_path = index_path
        self.metadata_path = metadata_path
        if self.index_path.exists() and self.metadata_path.exists():
            self.index = faiss.read_index(str(self.index_path))
            with open(self.metadata_path, 'r', encoding='utf-8') as f:
                self.metadata = json.load(f)
            print("✅ تم ربط قاعدة بيانات FAISS المحلية بنجاح.")
        else:
            self.index = None
            self.metadata = None

    def search(self, text_query, top_k=1, threshold=12.0): # Tightened threshold from 15.0 to 12.0
        if not self.index or not self.metadata:
            return None

        query_vector = self.embedder.encode([text_query]).astype('float32')
        distances, indices = self.index.search(query_vector, top_k)

        best_idx = indices[0][0]
        best_dist = distances[0][0]

        # If the distance is too large, the match is poor
        if best_idx != -1 and best_idx < len(self.metadata) and best_dist < threshold:
            return self.metadata[best_idx]

        return None

rag_engine = SmartSearchEngine(embedder, FAISS_INDEX_PATH, METADATA_PATH)
print("🎉 محرك البحث الذكي (RAG) جاهز للعمل مع فلترة أكثر صرامة للنتائج!")
```

```python
# ==========================================================
# 🚀 CELL 5: STEP 8 (RETRIEVE PRODUCT INFORMATION & FALLBACK)
# ==========================================================
from duckduckgo_search import DDGS

print("⏳ جاري بناء محرك استرجاع المعلومات (Step 8)...")

# --- Step 8.1: Define Web Search Fallback ---
class WebSearchEngine:
    def __init__(self): self.ddgs = DDGS()
    def search_product_live(self, query):
        try:
            print(f"🌍 جاري البحث الحي عن: '{query}'...")
            results = self.ddgs.text(keywords=query, region='ar-ar', max_results=3)
            if results: return f"اسم المنتج: {query}\nالوصف: {results[0].get('body', '')[:200]}"
        except Exception: pass
        return "لم يتم العثور على معلومات عبر الويب."

# --- Step 8.2: Define Retriever Class ---
class ProductInformationRetriever:
    def __init__(self, rag_engine, web_search_engine):
        self.rag_engine = rag_engine
        self.web_search_engine = web_search_engine

    def get_product_info(self, ocr_text):
        print(f"📦 [الخطوة 8] جاري البحث واسترجاع معلومات المنتج لـ: '{ocr_text}'")
        local_result = self.rag_engine.search(ocr_text)

        if isinstance(local_result, dict) and len(local_result) > 0:
            print("✅ تم العثور على المنتج في قاعدة البيانات المحلية (FAISS).")
            return f"اسم المنتج: {local_result.get('name', 'غير معروف')}\nالسعر: {local_result.get('price', 'غير متوفر')}\nالمكونات/الوصف: {local_result.get('description', 'لا توجد تفاصيل')}"

        print("🌐 المنتج غير موجود محلياً، جارِ تفعيل البحث الحي...")
        return self.web_search_engine.search_product_live(ocr_text)

# --- Step 8.3: Initialize Components ---
if 'web_search_engine' not in globals():
    web_search_engine = WebSearchEngine()

retriever = ProductInformationRetriever(rag_engine, web_search_engine)
print("🎉 الخطوة 8 (Retrieve Product Information) جاهزة ومربوطة بالكامل!")
```

```python
# ==========================================================
# 🚀 CELL 6: STEP 9 (BUILD RAG CONTEXT)
# ==========================================================

print("⏳ جاري إعداد صانع السياق (RAG Context Builder)...")

class RAGContextBuilder:
    def __init__(self):
        # تعليمات النظام (System Prompt) لتحديد شخصية الموديل
        self.system_prompt = (
            "أنت مساعد تسوق ذكي، دقيق، وخبير (AI Shopping Assistant). "
            "مهمتك هي الإجابة على استفسارات المستخدم بناءً على 'معلومات المنتج' المقدمة لك فقط. "
            "إذا كانت المعلومات غير كافية، اعتذر بلباقة ولا تقم بتأليف معلومات غير موجودة. "
            "قدم إجابتك باللغة العربية بأسلوب احترافي، منظم، ومختصر."
        )

    def build_prompt(self, ocr_text, product_info, user_query="ما هي تفاصيل هذا المنتج وهل تنصح به؟"):
        """
        الخطوة 9: تجميع وتغليف البيانات في قالب دقيق لإرساله إلى LLM (Qwen)
        """
        print(f"🧩 [الخطوة 9] جاري بناء وتغليف السياق للمنتج: '{ocr_text}'")

        # بناء القالب النهائي (Prompt Template)
        final_prompt = f"""{self.system_prompt}

### معلومات المنتج (التي تم استرجاعها):
{product_info}

### اسم المنتج المقروء من الصورة (OCR):
{ocr_text}

### استفسار المستخدم:
{user_query}

### إجابتك كمساعد تسوق:"""

        return final_prompt

# تهيئة صانع السياق
context_builder = RAGContextBuilder()
print("🎉 الخطوة 9 (Build RAG Context) جاهزة ومستعدة لتغليف البيانات!")
```

```python
# ==========================================================
# 🚀 CELL 7: STEP 10 (QWEN 2.5 LARGE LANGUAGE MODEL)
# ==========================================================
import os
import torch
from pathlib import Path

print("⏳ 1. جاري تجهيز بيئة الموديل الكبير (تسطيب المكتبات)...")
!pip install -q transformers accelerate bitsandbytes

from transformers import AutoModelForCausalLM, AutoTokenizer, BitsAndBytesConfig

# ---------------------------------------------------------
# 🟢 توجيه التحميل إلى Google Drive (Offline-Ready Hack)
# ---------------------------------------------------------
ROOT_DIR = Path("/content/drive/MyDrive/NTI tasks/AI Shopping Assistant")
HF_CACHE_DIR = ROOT_DIR / "models/hf_cache"
HF_CACHE_DIR.mkdir(parents=True, exist_ok=True)

# إجبار HuggingFace على تحميل وحفظ الأوزان في الدرايف للأبد
os.environ['HF_HOME'] = str(HF_CACHE_DIR)

class QwenAssistant:
    def __init__(self, model_id="Qwen/Qwen2.5-3B-Instruct"):
        print(f"⚙️ 2. جاري فحص وتحميل موديل {model_id} (هذا قد يستغرق وقتاً في المرة الأولى فقط)...")

        # إعدادات الضغط 4-bit لحماية الرامات وتسريع الأداء
        bnb_config = BitsAndBytesConfig(
            load_in_4bit=True,
            bnb_4bit_use_double_quant=True,
            bnb_4bit_quant_type="nf4",
            bnb_4bit_compute_dtype=torch.bfloat16
        )

        # تحميل الـ Tokenizer
        self.tokenizer = AutoTokenizer.from_pretrained(model_id)

        # تحميل الموديل الذكي مضغوطاً وموجهاً لكارت الشاشة
        self.model = AutoModelForCausalLM.from_pretrained(
            model_id,
            quantization_config=bnb_config,
            device_map="auto"
        )
        print("✅ تم تحميل موديل Qwen2.5 بنجاح وهو الآن متصل بكارت الشاشة!")

    def generate_answer(self, prompt_context):
        """
        الخطوة 10: توليد الإجابة النهائية بناءً على السياق (RAG)
        """
        print("🤖 Qwen يُفكر في الإجابة...")

        # تحويل النص المجمع إلى صيغة يفهمها الموديل
        messages = [
            {"role": "user", "content": prompt_context}
        ]

        text_input = self.tokenizer.apply_chat_template(
            messages,
            tokenize=False,
            add_generation_prompt=True
        )

        model_inputs = self.tokenizer([text_input], return_tensors="pt").to(self.model.device)

        # توليد الرد (مع تحديد طول الإجابة ومنع التكرار)
        generated_ids = self.model.generate(
            **model_inputs,
            max_new_tokens=256,
            temperature=0.3, # دقة عالية، إبداع أقل (لا نريد تأليف معلومات)
            repetition_penalty=1.1
        )

        # استخراج الإجابة الجديدة فقط (بدون تكرار السياق)
        generated_ids = [
            output_ids[len(input_ids):] for input_ids, output_ids in zip(model_inputs.input_ids, generated_ids)
        ]

        response = self.tokenizer.batch_decode(generated_ids, skip_special_tokens=True)[0]
        return response.strip()

# تهيئة الوحش الذكي
llm_engine = QwenAssistant()
print("🎉 الخطوة 10 جاهزة! Qwen2.5 مستعد للرد على أي استفسار.")
```
